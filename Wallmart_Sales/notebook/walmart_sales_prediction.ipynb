# Introduction


Sales forecasting is a critical aspect of retail business management, enabling organizations to optimize inventory, improve supply chain operations, plan staffing, and maximize profitability. Accurate predictions of future sales help retailers make informed decisions, reduce operational costs, and respond effectively to changing market demands.

In this project, a machine learning approach is used to predict Walmart's weekly sales based on historical sales data and several influencing factors, including store information, holiday events, temperature, fuel prices, Consumer Price Index (CPI), unemployment rate, and time-related features such as week and year.

The project follows a complete machine learning workflow, including data preprocessing, exploratory data analysis (EDA), feature engineering, model training, hyperparameter tuning, and performance evaluation. Multiple regression models were explored, with **XGBoost Regressor** selected as the final model due to its superior predictive performance.

The optimized XGBoost model achieved an **R² score of 0.9781**, **RMSE of 84,008**, and **MAE of 50,799**, demonstrating excellent accuracy in forecasting weekly sales. Feature importance analysis further provides insights into the factors that have the greatest influence on sales predictions.

This project demonstrates how advanced machine learning techniques can be applied to retail sales forecasting, providing a scalable and data-driven solution for business decision-making.

# Data & Imports

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

import kagglehub
from kagglehub import KaggleDatasetAdapter
```

```python
file_path = "Walmart_Sales.csv"

df = kagglehub.dataset_load(
  KaggleDatasetAdapter.PANDAS,
  "mikhail1681/walmart-sales",
  file_path,
)

df.to_csv('./data/Walmart_Sales.csv', index=False)

df = pd.read_csv("../data/Walmart_Sales.csv")
df.head()
```

```python
df.drop_duplicates(inplace=True)
```

# Data Cleaning

```python
df['Date'] = pd.to_datetime(df['Date'],format="%d-%m-%Y")

df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Week'] = df['Date'].dt.isocalendar().week.astype(int)
df['Quarter'] = df['Date'].dt.quarter


df.drop('Date',axis=1, inplace=True)
```

```python
df= df[['Year', 'Month','Week','Quarter','Store', 'Holiday_Flag', 'Temperature', 'Fuel_Price', 'CPI', 'Unemployment', 'Weekly_Sales']]
```

# Feature Selection

```python
target = df['Weekly_Sales']

features = df[['Year', 'Week', 'Store', 'Holiday_Flag', 'Temperature', 'Fuel_Price', 'CPI', 'Unemployment']]
```

## Import

```python
from sklearn.linear_model import LinearRegression,Lasso
from sklearn.model_selection import (KFold,cross_validate,RandomizedSearchCV)
from xgboost import XGBRegressor

from sklearn.metrics import (mean_squared_error,mean_absolute_error,r2_score)
import joblib
```

## Train-Test-Split

```python
x_train,x_test,y_train,y_test = train_test_split(features,target,test_size=0.2,random_state=42)
```

## Standardization

```python
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
## Saving Scaler
joblib.dump(scaler, "../model/scaler.pkl")
```

## Cross Validation

```python
cv = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

scoring = {
    'R2':'r2',
    'MAE':'neg_mean_absolute_error',
    'RMSE':'neg_root_mean_squared_error'
}
```

```python
def cross_validation(model,x,y):
    scores = cross_validate(
        model,
        x,
        y,
        cv = cv,
        scoring = scoring
    )

    print("Avg R2: ",scores["test_R2"].mean())
    print("Avg MAE: ",-scores["test_MAE"].mean())
    print("Avg RMSE: ",-scores["test_RMSE"].mean())
```

## Linear Regression

```python
lr = LinearRegression()
```

```python
cross_validation(
    lr,
    x_train_scaled,
    y_train,
)
```

## Lasso Regression

```python
ls = Lasso(
    alpha=0.1,
    random_state=42
)
```

```python
cross_validation(
    ls,
    x_train_scaled,
    y_train
)
```

## XG Boost

```python
xg = XGBRegressor(
    random_state=42
)
```

```python
cross_validation(
    xg,
    x_train,
    y_train
)
```

## Hyperparameter Tuning with RandomizedSearchCV

```python
params={
    "n_estimators":[100,200,300,400,500],
    "max_depth":[3,5,7,9],
    "learning_rate":[0.01,0.02,0.05,0.1,0.2],
    "subsample":[0.7,0.8,1.0],
    "colsample_bytree":[0.7,0.8,1.0]
}

random_search = RandomizedSearchCV(
    estimator = xg,
    param_distributions= params,
    n_iter= 20,
    scoring= 'r2',
    cv = 5,
    random_state=42,
    n_jobs=-1,
    verbose=2
)

random_search.fit(x_train, y_train)
```

```python
best_model = random_search.best_estimator_

print("Best Parameters: ")
print(random_search.best_params_)

print("\nBest Cross Validation R²: ")
print(random_search.best_score_)
```

## Feature Importance

```python
from xgboost import plot_importance
```

```python
plt.figure(figsize=(10,6))

plot_importance( best_model,importance_type="gain")

plt.tight_layout()
plt.savefig("../images/Feature_Importance.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
y_pred = best_model.predict(x_test)
```

## Save Model

```python
joblib.dump(best_model, "../model/XGBoost_model.pkl")
```

## Actual vs Predicted

```python
plt.figure(figsize=(8,6))

plt.scatter( y_test, y_pred, alpha=0.6)
plt.plot( [y_test.min(), y_test.max()],[y_test.min(), y_test.max()],color="red")

plt.xlabel("Actual Sales")
plt.ylabel("Predicted Sales")

plt.title("Actual vs Predicted")

plt.savefig("../images/Actual_vs_Predicted.png", dpi=300, bbox_inches="tight")
plt.show()
```

# Evaluation

```python
def evaluate(y_test,y_pred):
    print(f"MAE : {mean_absolute_error(y_test, y_pred)}")
    print(f"MSE : {mean_squared_error(y_test,y_pred)}")
    print(f"RMSE : {np.sqrt(mean_squared_error(y_test, y_pred))}")
    print(f"R² : {r2_score(y_test,y_pred)}")
```

```python
evaluate(y_test,y_pred)
```

### Residual Plot

```python
residuals = y_test-y_pred
```

```python
plt.figure(figsize=(8,6))
plt.scatter(y_pred,residuals,alpha=0.6)
plt.axhline(y=0,color='red',linestyle='--')
plt.xlabel("Predicted Sales")
plt.ylabel("Residuals")

plt.title("Residual Plot")
plt.savefig("../images/Residual_plot.png", dpi=300, bbox_inches='tight')
plt.show()
```

# Conclusion

The XGBoost Regressor demonstrated excellent performance in predicting Walmart weekly sales. After hyperparameter tuning using GridSearchCV, the model achieved an **R² score of 97.81%**, indicating that it explains nearly all of the variance in the target variable. The model also recorded a **Mean Absolute Error (MAE) of 50,798.88**, **Root Mean Squared Error (RMSE) of 84,007.97**, and a **Cross-Validation R² score of 97.63%**, confirming its strong predictive capability and good generalization on unseen data.

The **Actual vs Predicted** plot shows that most predictions closely align with the ideal diagonal line, indicating high prediction accuracy. The **Residual Plot** reveals that prediction errors are randomly distributed around zero without any significant pattern, suggesting that the model effectively captures the underlying relationship in the data with minimal bias. Feature importance analysis indicates that **Store**, **CPI**, and **Unemployment** are the most influential variables affecting weekly sales, while **Temperature**, **Fuel Price**, and **Holiday Flag** have relatively smaller contributions.

Overall, the developed XGBoost model is accurate, robust, and suitable for forecasting Walmart weekly sales. It can support inventory management, demand forecasting, and strategic business planning by providing reliable sales predictions.
